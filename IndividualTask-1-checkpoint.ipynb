```python
import pandas as pd

happiness = pd.read_csv("WHR_15_23.csv")
wdi = pd.read_csv("world_bank_development_indicators.csv")

print("Happiness columns:")
print(happiness.columns.tolist())
print("\nHappiness shape:",happiness.shape)
print(happiness.head())

print("\n\nWDI Columns:")
print(wdi.columns.tolist())
print("\nWDI shape:",wdi.shape)
print(wdi.head())
```

```python
wdi['year'] = pd.to_datetime(wdi['date']).dt.year

wdi_cols = ['country','year','GDP_current_US','inflation_annual%', 'life_expectancy_at_birth', 'gini_index', 'government_health_expenditure%', 
            'government_expenditure_on_education%', 'individuals_using_internet%','human_capital_index','political_stability_estimate',
           'rule_of_law_estimate', 'control_of_corruption_estimate', 'population']

wdi_subset = wdi[wdi_cols]

merged = pd.merge(happiness, wdi_subset, on=['country', 'year'], how='inner')
print("Merged Columns:",merged.shape)
print("Countries lost:")
print(set(happiness['country'].unique()) - set(merged['country'].unique()))
```

```python
rename_dict = {
    'Russia': 'Russian Federation', 'South Korea': 'Korea, Rep.', 'Egypt': 'Egypt, Arab Rep.', 'Iran': 'Iran, Islamic Rep.', 'Syria': 'Syrian Arab Republic',
    'Yemen': 'Yemen, Rep.', 'Venezuela': 'Venezuela, RB', 'Kyrgyzstan': 'Kyrgyz Republic', 'Laos': 'Lao PDR', 'Congo (Brazzaville)': 'Congo, Rep.',
    'Congo (Kinshasa)': 'Congo, Dem. Rep.', 'Gambia': 'Gambia, The','Ivory Coast': "Cote d'Ivoire", 'Slovakia': 'Slovak Republic','Swaziland': 'Eswatini', 
    'Hong Kong S.A.R. of China': 'Hong Kong SAR, China', 'State of Palestine': 'West Bank and Gaza','Palestinian Territories': 'West Bank and Gaza',
}

happiness_fixed = happiness.copy()
happiness_fixed['country'] = happiness_fixed['country'].replace(rename_dict)

wdi_countries = set(wdi_subset['country'].unique())
for old, new in rename_dict.items():
    status = "found" if new in wdi_countries else "not found"
    print(f"{old} -> {new}: {status}")
```

```python
merged = pd.merge(happiness_fixed, wdi_subset, on=['country', 'year'], how='inner')
print("\nNew merged:",merged.shape)
print("\nUnmatched ones:")
print(set(happiness_fixed['country'].unique()) - set(merged['country'].unique()))
```

```python
merged['happiness_tier'] = pd.qcut(merged['happiness_score'], q=3, labels=['Low', 'Medium', 'High'])
print(merged['happiness_tier'].value_counts())
```

```python
wdi_features = ['GDP_current_US', 'inflation_annual%', 'life_expectancy_at_birth', 'gini_index','government_health_expenditure%',
                'government_expenditure_on_education%', 'individuals_using_internet%','human_capital_index','political_stability_estimate',
               'rule_of_law_estimate','control_of_corruption_estimate','population']

missing_one = merged[wdi_features].isnull().mean().sort_values(ascending=False) * 100
print(missing_one)
```

```python
updated_features = ['GDP_current_US', 'inflation_annual%', 'life_expectancy_at_birth', 'government_health_expenditure%', 'government_expenditure_on_education%',
                   'individuals_using_internet%', 'political_stability_estimate', 'rule_of_law_estimate', 'control_of_corruption_estimate', 'population']

for col in updated_features:
    median_val = merged[col].median()
    merged[col] = merged[col].fillna(median_val)

print(merged[updated_features].isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

x = merged[updated_features]
y = merged['happiness_tier']

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 38)

print("Train size:", x_train.shape)
print("Test size:", x_test.shape)
print('\nTrain tier balance:')
print(y_train.value_counts(normalize=True))
print("\nTest tier balance:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

dt = DecisionTreeClassifier(max_depth=5, random_state=39)
dt.fit(x_train, y_train)
dt_pred = dt.predict(x_test)

print("Decision Tree Results: ")
print("Accuracy:", accuracy_score(y_test, dt_pred))
print("F1 score:",f1_score(y_test, dt_pred, average='macro'))
print("Classification Report:")
print(classification_report(y_test, dt_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, dt_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=200, random_state=39)
rf.fit(x_train, y_train)
rf_pred = rf.predict(x_test)

print("Random Forest Classifier Results:")
print("Accuracy:", accuracy_score(y_test, rf_pred))
print("F1 score:",f1_score(y_test, rf_pred, average='macro'))
print("Classification Report:")
print(classification_report(y_test, rf_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, rf_pred))
```

```python
dt_importance = pd.Series(dt.feature_importances_, index=updated_features).sort_values(ascending=False)
rf_importance = pd.Series(rf.feature_importances_, index=updated_features).sort_values(ascending=False)

print("Decision Tree - Feature Importance:")
print(dt_importance)
print("\nRandom Forest - Feature Importance:")
print(rf_importance)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

importance_df = pd.DataFrame({
    'Decision Tree': dt_importance, 'Random Forest': rf_importance
}).sort_values('Random Forest', ascending=True)

fig, ax = plt.subplots(figsize=(10, 7))
importance_df.plot(kind='barh', ax=ax, color = ['#52ABDD', '#DD5266'])
ax.set_xlabel('Feature Importance Score')
ax.set_title('Feature importance: Decision Tree vs Random Forest')
ax.legend(title='Model')
plt.tight_layout()
plt.savefig('feat_imp_comp.png', dpi=300)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 6))

sns.heatmap(confusion_matrix(y_test, dt_pred), annot=True, fmt='d', cmap='Reds',
            xticklabels=['High', 'Low', 'Medium'], yticklabels=['High', 'Low', 'Medium'],
            ax=axes[0])
axes[0].set_title('Decision Tree - Confusion Matrix')
axes[0].set_xlabel('Predicted')
axes[0].set_ylabel('Actual')

sns.heatmap(confusion_matrix(y_test, rf_pred), annot=True, fmt='d', cmap='Blues',
            xticklabels=['High', 'Low', 'Medium'], yticklabels=['High', 'Low', 'Medium'],
            ax=axes[1])
axes[1].set_title('Random Forest - Confusion Matrix')
axes[1].set_xlabel('Predicted')
axes[1].set_ylabel('Actual')

plt.tight_layout()
plt.savefig('conf_matr.png', dpi=300)
plt.show()
```

```python
comparison_df = pd.DataFrame({
    'Decision Tree': [accuracy_score(y_test, dt_pred), f1_score(y_test, dt_pred, average='macro')],
    'Random Forest': [accuracy_score(y_test, rf_pred), f1_score(y_test, rf_pred, average='macro')]
}, index=['Accuracy', 'F1 (macro)'])

fig, ax = plt.subplots(figsize=(8, 6))
comparison_df.plot(kind='bar', ax=ax, color=['#8452DD', '#DD8452'], rot=0)
ax.set_ylabel('Score')
ax.set_ylim(0, 1)
ax.set_title('Comparison of Model Performance')
plt.tight_layout()
plt.savefig('model_comp.png', dpi=300)
plt.show()
```
